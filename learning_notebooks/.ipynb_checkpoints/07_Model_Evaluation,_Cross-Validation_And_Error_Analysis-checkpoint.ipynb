```python
import joblib
df = joblib.load("../learning_notebooks/data/df.pkl")
rf_model = joblib.load("../learning_notebooks/model/rf_model.pkl")
```

```python
df
```

```python
X=df.drop('Loan_Status_Y',axis=1)
y=df['Loan_Status_Y']
```

```python
# cross validation
```

```python
# k fold code 
from sklearn.model_selection import cross_val_score
score=cross_val_score(
    rf_model, 
    X,
    y,
    cv=5, 
    scoring='f1'
)
```

```python
score
```

```python
score.mean()
```

```python
from sklearn.model_selection import StratifiedKFold
skf=StratifiedKFold(n_splits=5,shuffle=True,random_state=42)
```

```python
skf
```

```python
# learning curve
```

```python
# plot learning curve
```

```python
from sklearn.model_selection import learning_curve
import numpy as np
train_sizes,train_score,val_score=learning_curve(
    rf_model, 
    X, 
    y, 
    cv=5, 
    scoring='f1', 
    train_sizes=np.linspace(0.1,1.0,5)
)
train_mean=train_score.mean(axis=1) 
val_mean=val_score.mean(axis=1)
```
